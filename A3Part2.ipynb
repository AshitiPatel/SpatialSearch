```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer, set_seed

# CRITERIA 3: Deterministic Output
# Setting a manual seed ensures that even though we are generating text,
# the model will make the exact same choices every time this cell is run.
set_seed(42)

# Load the tokenizer and the model (we are using the base 'gpt2' model)
print("Downloading and loading GPT-2... (Takes a few seconds)\n")
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")
model = GPT2LMHeadModel.from_pretrained("gpt2")

# Create a starting prompt to give the model some context
prompt_text = "Commuting home on the 5:30 PM GO Train, she tuned out the chatter around her and buried herself in 'The Predator' by RuNyx. Just as the train approached her station,"
input_ids = tokenizer.encode(prompt_text, return_tensors='pt')

# Generate the story!
# CRITERIA 1 & 2: Length and Repetition
print("Generating the story...\n")
output = model.generate(
    input_ids,
    max_length=200,             # Ensures the output is well over 100 words
    no_repeat_ngram_size=3,     # Prevents the model from repeating any 3-word phrases
    do_sample=True,             # Allows for creative text generation
    top_k=50,                   # Restricts vocabulary to top 50 most likely words (improves quality)
    top_p=0.95,                 # Nucleus sampling (improves quality)
    pad_token_id=tokenizer.eos_token_id # Clears a warning about padding tokens
)

# Decode the generated tokens back into human-readable text
story = tokenizer.decode(output[0], skip_special_tokens=True)

# Print the final story and a word count check for the TA
print("--- GENERATED STORY ---")
print(story)
print("-----------------------\n")

word_count = len(story.split())
print(f"Total Word Count: {word_count} words (Requirement: >100)")
```

```python
# Write the story variable to a text file
with open("A3Part2.txt", "w") as file:
    file.write(story)

# Prompt Colab to download the file to local machine
from google.colab import files
files.download("A3Part2.txt")
```
